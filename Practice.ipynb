```python
%%sql

SELECT department, count(department) as Number
FROM pratice.students
GROUP BY
department
```

```python
%%sql

SELECT department, avg(marks) as Number
FROM pratice.students
GROUP BY
department
```

```python
%%sql

SELECT name, marks,age,
    case
        when marks >= 50 then 'Pass'
        else 'Fail'
    end as Result
FROM pratice.students
```

```python
%%sql

SELECT name, marks,
    case
        when marks >= 90 then 'A'
        when marks between 75 and 89 then 'B'
        when marks between 50 and 74 then 'C'
        else 'F'
    end as Grade
FROM pratice.students
```

```python
%%sql

SELECT name,age, marks,
    case
        when marks >= 90 then 'Excellent'
        when marks >= 75 then 'Good'
        else 'Needs Improvement'
    end as result
FROM pratice.students
```
