```python
import torch

# Starting weight
w = torch.tensor(1.0, requires_grad=True)

# Input and correct answer
x = torch.tensor(1.0)
y = torch.tensor(2.0)

# Forward pass
y_hat = w * x
s = y_hat - y
L = s ** 2

print("Prediction:", y_hat.item())
print("Error:", s.item())
print("Loss:", L.item())

# Backward pass
L.backward()

print("Gradient:", w.grad.item())
```
