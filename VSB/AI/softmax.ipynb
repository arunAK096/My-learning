```python
import torch
import torch.nn as nn
import numpy as np
```

```python
def softmax(x):
    return np.exp(x) / np.sum(np.exp(x), axis=0)

x = np.array([2.0, 1.0, 0.1])
print(softmax(x))

x= torch.tensor([2.0, 1.0, 0.1])
print(torch.softmax(x, dim=0))
```

```python
def CrossEntropyLoss(output, correct):
    return -np.sum(correct * np.log(output))

correct = np.array([1,0,0])

output_good = np.array([0.7, 0.2, 0.1])
output_bad = np.array([0.1, 0.3, 0.6])

l1 = CrossEntropyLoss(output_good, correct)
l2 = CrossEntropyLoss(output_bad, correct)

print(f"Loss for good output: {l1:.4f}")
print(f"Loss for bad output: {l2:.4f}")
```

```python
loss = nn.CrossEntropyLoss()

y = torch.tensor([2,0,1])

y_prect_good = torch.tensor([[2.0, 1.0, 0.1], [1.0, 2.0, 0.1], [0.1, 0.1, 2.0]])
y_prect_bad = torch.tensor([[0.1, 0.3, 0.6], [0.3, 0.1, 0.6], [0.6, 0.1, 0.3]])

l1 = loss(y_prect_good, y)
l2 = loss(y_prect_bad, y)   

print(f"l1: {l1.item():.4f}")
print(f"l2: {l2.item():.4f}")

_, predicted1 = torch.max(y_prect_good, 1)
_, predicted2 = torch.max(y_prect_bad, 1)

print(predicted1)
print(predicted2)
```
