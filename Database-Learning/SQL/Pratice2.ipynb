```python
%%sql

CREATE TABLE sql_practice.Students(
    student_id INT PRIMARY KEY,
    student_name VARCHAR(50),
    age INT,
    city VARCHAR(50),
    course_id INT
);

CREATE TABLE sql_practice.Courses (
    course_id INT PRIMARY KEY,
    course_name VARCHAR(50),
    fee INT
);

CREATE TABLE sql_practice.Marks (
    mark_id INT PRIMARY KEY,
    student_id INT,
    subject VARCHAR(50),
    marks INT
);
```
